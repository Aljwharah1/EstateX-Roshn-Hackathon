```python
import pandas as pd
df = pd.read_excel('Finalized_Data.xlsx')
```

```python
# Filter for North apartments with budget 500k-1M SAR
location = 'شمال'
property_type = 'apartment'
min_budget = 500000
max_budget = 1000000
filtered = df[(df['location'] == location) & (df['property_type'] == property_type) & (df['price_sar'] >= min_budget) & (df['price_sar'] <= max_budget)]
print(f'Properties matching criteria: {len(filtered)}')
if len(filtered) > 0:
    display(filtered[['location', 'property_type', 'price_sar', 'district']].head())
else:
    print('No properties found matching all criteria.')
```
